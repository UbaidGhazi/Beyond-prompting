````python
# Environment setup: Add repository root to path & import shared helper
import sys
from pathlib import Path

current_path = Path('.').resolve()
repo_root = current_path if (current_path / 'src').exists() else current_path.parent.parent
if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))

from src.ai_helper import ask_gemini
print('[PASS] Environment configured and ready to connect to Google Gemini!')
````

````python
# 1. Vague Prompt
vague_prompt = 'Suggest an activity for university students.'

# 2. Specific Prompt with campus constraints
specific_prompt = (
    'Suggest a 30-minute icebreaker activity for 40 first-semester '
    'Computer Science students at COMSATS University Islamabad (Wah Campus). '
    'The activity must require zero budget, work inside CS Lab 03, '
    'and help students learn each other names and coding interests.'
)

print('--- [RESPONSE 1: VAGUE INPUT] ---')
print(ask_gemini(vague_prompt).strip())

print('\n--- [RESPONSE 2: SPECIFIC INPUT] ---')
print(ask_gemini(specific_prompt).strip())
````

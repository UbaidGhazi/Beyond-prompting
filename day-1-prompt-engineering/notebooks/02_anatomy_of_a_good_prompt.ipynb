````python
# Environment setup: Add repository root to path & import shared helper
import sys
from pathlib import Path

current_path = Path('.').resolve()
repo_root = current_path if (current_path / 'src').exists() else current_path.parent.parent
if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))

from src.ai_helper import ask_gemini
print('[PASS] Environment configured and ready to connect to Google Gemini!')
````

````python
role = 'Role: You are the Lead Community Organizer for GDGoC CUI Wah.'
task = 'Task: Write an energetic WhatsApp announcement for a beginner workshop.'
context = 'Context: Git & GitHub Essentials, Friday Nov 27 @ 2 PM in CS Lab 05, CUI Wah.'
constraints = 'Constraints: Max 100 words, friendly tone, include 3 takeaway bullets.'
format_str = 'Format: [Headline] -> [Hook] -> Key Details -> [3 Takeaways] -> [CTA]'

prompt = f'{role}\n{task}\n{context}\n{constraints}\n{format_str}'
print('--- [GENERATING 5-PILLAR ANNOUNCEMENT] ---')
print(ask_gemini(prompt).strip())
````

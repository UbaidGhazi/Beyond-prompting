````python
# Environment setup: Add repository root to path & import shared helper
import sys
from pathlib import Path

current_path = Path('.').resolve()
repo_root = current_path if (current_path / 'src').exists() else current_path.parent.parent
if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))

from src.ai_helper import ask_gemini
print('[PASS] Environment configured and ready to connect to Google Gemini!')
````

````python
test_msg = 'Salam! Can a local software house set up a branded booth at the hackathon?'

few_shot_prompt = (
    'Classify the message into: [REGISTRATION], [VENUE_INQUIRY], [SPONSORSHIP].\n\n'
    'Example 1: "Is Lab 03 on the 1st floor?" -> [VENUE_INQUIRY]\n'
    'Example 2: "I submitted the form but got no email." -> [REGISTRATION]\n'
    'Example 3: "We offer PKR 30,000 for swag branding." -> [SPONSORSHIP]\n\n'
    f'Message: "{test_msg}"\n'
    'Category:'
)

print('--- [FEW-SHOT CLASSIFICATION RESULT] ---')
print(ask_gemini(few_shot_prompt, temperature=0.1).strip())
````

````python
# Environment setup: Add repository root to path & import shared helper
import sys
from pathlib import Path

current_path = Path('.').resolve()
repo_root = current_path if (current_path / 'src').exists() else current_path.parent.parent
if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))

from src.ai_helper import ask_gemini
print('[PASS] Environment configured and ready to connect to Google Gemini!')
````

````python
# Step 1: Objectives
p1 = 'Define 3 learning objectives for a 1-day CUI Wah hackathon (SpeedCode). Under 60 words.'
res1 = ask_gemini(p1, temperature=0.5)
print('>> Step 1 (Objectives):\n' + res1.strip() + '\n')

# Step 2: Schedule using Step 1
p2 = f'Using these objectives:\n{res1}\nDraft a 4-milestone schedule for CS Lab 01.'
res2 = ask_gemini(p2, temperature=0.5)
print('>> Step 2 (Schedule):\n' + res2.strip() + '\n')

# Step 3: Broadcast using Step 2
p3 = f'Using this schedule:\n{res2}\nWrite a 50-word WhatsApp student announcement.'
res3 = ask_gemini(p3, temperature=0.7)
print('>> Step 3 (Broadcast):\n' + res3.strip())
````

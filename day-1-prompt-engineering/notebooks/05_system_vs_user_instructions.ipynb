````python
# Environment setup: Add repository root to path & import shared helper
import sys
from pathlib import Path

current_path = Path('.').resolve()
repo_root = current_path if (current_path / 'src').exists() else current_path.parent.parent
if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))

from src.ai_helper import ask_gemini
print('[PASS] Environment configured and ready to connect to Google Gemini!')
````

````python
user_msg = 'Explain what a Python for loop is and when to use it.'

# Persona A: Strict Professor
sys_prof = 'You are a strict CS professor. Explain in exactly 2 formal, academic sentences.'
print('--- [PROFESSOR PERSONA] ---')
print(ask_gemini(user_msg, system_instruction=sys_prof).strip())

# Persona B: Roman Urdu Senior Mentor
sys_urdu = 'Aap CUI Wah ke friendly senior ho. Roman Urdu mein canteen ki misaal de kar samjhao.'
print('\n--- [ROMAN URDU MENTOR PERSONA] ---')
print(ask_gemini(user_msg, system_instruction=sys_urdu).strip())
````

````python
# Environment setup: Add repository root to path & import shared helper
import sys
from pathlib import Path

current_path = Path('.').resolve()
repo_root = current_path if (current_path / 'src').exists() else current_path.parent.parent
if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))

from src.ai_helper import ask_gemini
print('[PASS] Environment configured and ready to connect to Google Gemini!')
````

````python
import json

prompt = (
    'Generate a JSON object for a CUI Wah coding workshop.\n'
    'Include keys: "event_name" (str), "date" (str), "venue" (str), "prerequisites" (list).\n'
    'Output valid JSON only.'
)

raw_json = ask_gemini(prompt, response_mime_type='application/json', temperature=0.2)
print('Raw JSON string returned from API:\n' + raw_json.strip() + '\n')

data = json.loads(raw_json)
print(f'Successfully parsed into Python dict! Type: {type(data)}')
print(f'Workshop: {data.get("event_name")} | Venue: {data.get("venue")}')
````

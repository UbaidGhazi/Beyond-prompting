````python
# Environment setup: Add repository root to path & import shared helper
import sys
from pathlib import Path

current_path = Path('.').resolve()
repo_root = current_path if (current_path / 'src').exists() else current_path.parent.parent
if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))

from src.ai_helper import ask_gemini
print('[PASS] Environment configured and ready to connect to Google Gemini!')
````

````python
q = 'Where and at what time is the AI Robotics Workshop happening at CUI Wah?'

# Call 1: Without context
print('--- [CALL 1: NO CONTEXT] ---')
print(ask_gemini(q).strip()[:180] + '...\n')

# Call 2: With grounded context
notice = 'Robotics Workshop: Nov 18, 11:30 AM in Electrical Lab 02, Block 1, CUI Wah.'
prompt = f'Use ONLY this notice to answer:\n{notice}\n\nQuestion: {q}'
print('--- [CALL 2: WITH GROUNDED CONTEXT] ---')
print(ask_gemini(prompt).strip())
````

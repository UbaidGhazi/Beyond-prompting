````python
# Environment setup: Add repository root to path & import shared helper
import sys
from pathlib import Path

current_path = Path('.').resolve()
repo_root = current_path if (current_path / 'src').exists() else current_path.parent.parent
if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))

from src.ai_helper import ask_gemini
print('[PASS] Environment configured and ready to connect to Google Gemini!')
````

````python
question = 'What is the registration deadline and fee for the upcoming hackathon?'

ctx_a = 'Event: SpeedCode Hackathon 2026. Fee: PKR 500. Deadline: Nov 16. Venue: SAC.'
ctx_b = 'Event: DataHack 2026. Fee: Free (sponsored). Deadline: Dec 05. Venue: Auditorium.'

print('--- [EVALUATING WITH CONTEXT A] ---')
print(ask_gemini(f'Context:\n{ctx_a}\nQuestion: {question}').strip())

print('\n--- [EVALUATING WITH CONTEXT B] ---')
print(ask_gemini(f'Context:\n{ctx_b}\nQuestion: {question}').strip())
````

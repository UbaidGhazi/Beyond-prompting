````python
# Environment setup: Add repository root to path & import shared helper
import sys
from pathlib import Path

current_path = Path('.').resolve()
repo_root = current_path if (current_path / 'src').exists() else current_path.parent.parent
if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))

from src.ai_helper import ask_gemini
print('[PASS] Environment configured and ready to connect to Google Gemini!')
````

````python
NOTICES = [
    {'id': 'N1', 'title': 'Futsal Cup', 'text': 'Futsal starts Dec 02 at campus grounds.'},
    {'id': 'N2', 'title': 'Python Workshop', 'text': 'Python logic workshop in Lab 03 on Nov 05 (Free).'},
    {'id': 'N3', 'title': 'Library Notice', 'text': 'Library study room relocated to Seminar Room A.'}
]

def filter_notes(query, notes):
    words = [w.lower() for w in query.split() if len(w) > 3]
    return [n for n in notes if any(w in (n['title'] + ' ' + n['text']).lower() for w in words)]

query = 'Where is the Python workshop happening?'
selected = filter_notes(query, NOTICES)
print(f'Retrieved {len(selected)} matching record(s): {[s["id"] for s in selected]}')

ctx = selected[0]['text'] if selected else 'No record'
print(ask_gemini(f'Context: {ctx}\nQuestion: {query}').strip())
````

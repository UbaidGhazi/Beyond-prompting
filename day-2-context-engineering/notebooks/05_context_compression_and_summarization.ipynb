````python
# Environment setup: Add repository root to path & import shared helper
import sys
from pathlib import Path

current_path = Path('.').resolve()
repo_root = current_path if (current_path / 'src').exists() else current_path.parent.parent
if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))

from src.ai_helper import ask_gemini
print('[PASS] Environment configured and ready to connect to Google Gemini!')
````

````python
policy_doc = (
    'CUI WAH COMPUTING LAB POLICY 2026: '
    '1. Students must wear university ID lanyards at all times in Labs 01-06. '
    '2. Backpacks must be placed on wooden storage racks near entry. '
    '3. Strictly no food items or chips permitted near terminals. '
    '4. Evening hackathons past 5 PM require a 24-hour approved gate pass.'
)

# Call 1: Compress
summary = ask_gemini(f'Condense this lab policy into 3 concise bullets:\n{policy_doc}')
print('--- [GENERATED COMPACT SUMMARY] ---')
print(summary.strip() + '\n')

# Call 2: QA
q = 'Can I eat chips at my computer desk in CS Lab 03?'
print('--- [ANSWER FROM SUMMARY] ---')
print(ask_gemini(f'Context:\n{summary}\nQuestion: {q}').strip())
````

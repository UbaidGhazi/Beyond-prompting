````python
# Environment setup: Add repository root to path & import shared helper
import sys
from pathlib import Path

current_path = Path('.').resolve()
repo_root = current_path if (current_path / 'src').exists() else current_path.parent.parent
if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))

from src.ai_helper import ask_gemini
print('[PASS] Environment configured and ready to connect to Google Gemini!')
````

````python
# Turn 1: Save student state in Python
app_memory = {'student_name': 'Zainab', 'semester': 1, 'level': 'Beginner'}
print(f'Stored Application Memory: {app_memory}')

# Turn 2: Inject state into follow-up
query = 'What workshop should I register for first?'
prompt = (
    f'Student Profile: {app_memory}\n'
    f'Available: [Python for Beginners, Advanced Kubernetes, C++ Data Structures]\n'
    f'Question: {query}\n'
    'Recommend the best workshop and address the student by name.'
)

print('--- [STATE-AWARE PERSONALIZED RESPONSE] ---')
print(ask_gemini(prompt).strip())
````

````python
# Environment setup: Add repository root to path & import shared helper
import sys
from pathlib import Path

current_path = Path('.').resolve()
repo_root = current_path if (current_path / 'src').exists() else current_path.parent.parent
if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))

from src.ai_helper import ask_gemini
print('[PASS] Environment configured and ready to connect to Google Gemini!')
````

````python
DOCS = [
    {'id': '1', 'title': 'Git Workshop', 'text': 'Git workshop is on Nov 27 in CS Lab 05 at 2 PM.'},
    {'id': '2', 'title': 'SpeedCode', 'text': 'SpeedCode hackathon is on Nov 20 in the SAC. Fee PKR 500.'}
]

def mini_rag(query):
    matched = [d['text'] for d in DOCS if any(w in d['text'].lower() for w in query.lower().split() if len(w)>3)]
    ctx = '\n'.join(matched) if matched else 'No records found'
    prompt = f'Answer ONLY from context. If not found, say "Information unavailable".\nContext: {ctx}\nQuery: {query}'
    return ask_gemini(prompt, temperature=0.1).strip()

print('Test 1 (Answerable): ' + mini_rag('When is the Git session?'))
print('Test 2 (Missing):    ' + mini_rag('What time does the swimming pool open?'))
````

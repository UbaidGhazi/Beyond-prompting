````python
# Environment setup: Add repository root to path & import shared helper
import sys
from pathlib import Path

current_path = Path('.').resolve()
repo_root = current_path if (current_path / 'src').exists() else current_path.parent.parent
if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))

from src.ai_helper import ask_gemini
print('[PASS] Environment configured and ready to connect to Google Gemini!')
````

````python
# Simulated Python Tool
def get_timetable(code):
    table = {'CSC101': {'room': 'CS Lab 02', 'time': '09:00 AM', 'teacher': 'Engr. Ali'}}
    return table.get(code, {'error': 'Class not found'})

course = 'CSC101'
live_data = get_timetable(course)
print(f'Tool Returned Live Data: {live_data}\n')

prompt = (
    f'Live Campus System Data: {live_data}\n'
    f'Student Question: Where is my {course} class this morning?\n'
    'Answer in a friendly, conversational sentence.'
)

print(ask_gemini(prompt).strip())
````

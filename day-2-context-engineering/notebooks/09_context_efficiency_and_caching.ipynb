````python
# Environment setup: Add repository root to path & import shared helper
import sys
from pathlib import Path

current_path = Path('.').resolve()
repo_root = current_path if (current_path / 'src').exists() else current_path.parent.parent
if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))

from src.ai_helper import ask_gemini
print('[PASS] Environment configured and ready to connect to Google Gemini!')
````

````python
import hashlib, time

CACHE = {}
def smart_ask(prompt):
    key = hashlib.md5(prompt.strip().lower().encode()).hexdigest()
    t0 = time.perf_counter()
    if key in CACHE:
        return CACHE[key], 'CACHE_HIT', time.perf_counter() - t0
    res = ask_gemini(prompt, temperature=0.1)
    CACHE[key] = res
    return res, 'CACHE_MISS', time.perf_counter() - t0

q = 'What are the prerequisites for the Git workshop at CUI Wah?'
res1, stat1, t1 = smart_ask(q)
print(f'Call 1: [{stat1}] in {t1:.3f}s')

res2, stat2, t2 = smart_ask(q)
print(f'Call 2: [{stat2}] in {t2:.6f}s (Instantaneous!)')
````

````python
# Environment setup: Add repository root to path & import shared helper
import sys
from pathlib import Path

current_path = Path('.').resolve()
repo_root = current_path if (current_path / 'src').exists() else current_path.parent.parent
if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))

from src.ai_helper import ask_gemini
print('[PASS] Environment configured and ready to connect to Google Gemini!')
````

````python
# Stage 1 & 2: Rules and Context
event_info = {'title': 'Python Workshop', 'date': 'Nov 05', 'venue': 'CS Lab 03'}
rules = 'You are a campus assistant. Write a short WhatsApp broadcast containing Event, Date, and Venue.'
prompt = f'Event Data: {event_info}\nDraft announcement.'

# Stage 3: Call Model
response = ask_gemini(prompt, system_instruction=rules, temperature=0.3)

# Stage 4 & 5: Inspect and Deterministically Validate with Python
checks = {
    'has_venue': 'lab 03' in response.lower(),
    'has_date': 'nov' in response.lower(),
    'length_valid': len(response.split()) <= 80
}
print(f'Deterministic Validation Report: {checks}')

# Stage 6: Deliver
if all(checks.values()):
    print('\n>> [PIPELINE SUCCESS] Approved Announcement:\n' + response.strip())
else:
    print('\n>> [VALIDATION FAILED] Re-routing to retry...')
````
